```python
# Gộp dữ liệu các bảng bằng left join
# Import mode đúng cách
from scipy.stats import mode
import numpy as np

def get_mode(x):
    # Loại bỏ các giá trị không mong muốn
    filtered = x[~x.isin(["Unknown", "UnknownIncome"])]

    # Nếu sau khi lọc vẫn còn dữ liệu, trả về mode
    if not filtered.empty:
        return filtered.mode().iloc[0] if not filtered.mode().empty else "Unknown"
    else:
        # Nếu không còn dữ liệu nào khác, trả về "Unknown"
        return "Unknown"
```

```python
df_2_mode = Brand_Image_Cleaned.groupby(['ID', 'Year', 'City'], as_index=False)[['Attribute']].agg(get_mode)
df_3_mode = Companion_Cleaned.groupby(['ID', 'Year', 'City'], as_index=False)[['Companion#group']].agg(get_mode)
df_4_mode = Day_of_week_Cleaned.groupby(['ID', 'Year', 'City'], as_index=False)[['Dayofweek']].agg(get_mode)
df_5_mode = Day_part_cleaned.groupby(['ID', 'Year', 'City'], as_index=False)[['Daypart']].agg(get_mode)
df_6_mode = Need_Sate_Day_cleaned.groupby(['ID', 'Year', 'City'], as_index=False)[['Needstates']].agg(get_mode)

df_temp = HL_Customers \
    .merge(df_2_mode, on=['ID', 'Year', 'City'], how='left') \
    .merge(df_3_mode, on=['ID', 'Year', 'City'], how='left') \
    .merge(df_4_mode, on=['ID', 'Year', 'City'], how='left') \
    .merge(df_5_mode, on=['ID', 'Year', 'City'], how='left') \
    .merge(df_6_mode, on=['ID', 'Year', 'City'], how='left')
```

```python
_cluster = df_temp_1[['ID', 'Fre#visit', 'Spending', 'NPS#P3M','Companion#group','Daypart', 'NeedstateGroup']]
cols = ['NPS#P3M']
_cluster[cols] = _cluster[cols].astype(float)
_onemoretime = _cluster[['Fre#visit', 'Spending', 'NPS#P3M', 'Companion#group', 'Daypart', 'NeedstateGroup']]
_onemoretime[cols] = _onemoretime[cols].astype(int)

#Get list of categorical variables
s = (_onemoretime.dtypes == 'object')
object_cols = list(s[s].index)

print("Categorical variables in the dataset:", object_cols)

# Label Encoding the object dtypes.
encoders = {}
for i in object_cols:
    LE = LabelEncoder()
    _onemoretime[i] = _onemoretime[[i]].apply(LE.fit_transform)
    encoders[i] = LE  # Lưu lại encoder để biết mapping

Clusters = _onemoretime.copy()

# Scale cột
scaler = StandardScaler()
_onemoretime = scaler.fit_transform(_onemoretime)
```

```python
#Initiating PCA to reduce dimentions aka features to 3
pca = PCA(n_components=3)
pca.fit(_onemoretime)
PCA_ds = pd.DataFrame(pca.transform(_onemoretime), columns=(["col1","col2", "col3"]))
PCA_ds.describe().T
```

```python
# A 3D Projection Of Data In The Reduced Dimension
x =PCA_ds["col1"]
y =PCA_ds["col2"]
z =PCA_ds["col3"]
# To plot
fig = plt.figure(figsize=(10,8))
ax = fig.add_subplot(111, projection="3d")
ax.scatter(x,y,z, c="maroon", marker="o" )
ax.set_title("A 3D Projection Of Data In The Reduced Dimension")
plt.show()
```

```python
# Quick examination of elbow method to find numbers of clusters to make.
print('Elbow Method to determine the number of clusters to be formed:')
Elbow_M = KElbowVisualizer(KMeans(), k=10)
Elbow_M.fit(PCA_ds)
Elbow_M.show()
```

```python
#Initiating the Agglomerative Clustering model
AC = AgglomerativeClustering(n_clusters=5)
# fit model and predict clusters
yhat_AC = AC.fit_predict(PCA_ds)
PCA_ds["Clusters"] = yhat_AC
#Adding the Clusters feature to the orignal dataframe.
_cluster["Clusters"]= yhat_AC
```

```python
#Plotting the clusters
fig = plt.figure(figsize=(10,8))
ax = plt.subplot(111, projection='3d', label="bla")
ax.scatter(x, y, z, s=40, c=PCA_ds["Clusters"], marker='o', cmap = 'viridis' )
ax.set_title("The Plot Of The Clusters")
plt.show()
```

```python
from sklearn.metrics import silhouette_score
score = silhouette_score(PCA_ds, yhat_AC)
print(f"Silhouette Score của Agglomerative Clustering: {score:.3f}")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import shap
import warnings
warnings.filterwarnings("ignore")

# ============ BƯỚC 1: TẠO CỜ CHURN ============
df_temp_1['churn_flag'] = np.where(
    (df_temp_1['P3M'] == 'Highlands Coffee') & (df_temp_1['P1M'] == 'No'),
    1, 0
)

Clusters = Clusters.merge(_cluster[['ID']], left_index=True, right_index=True)

# Nếu 'Cluster' chưa có thì merge
if 'Clusters' not in df_temp_1.columns:
    df_temp_1 = df_temp_1.merge(
        Clusters[['ID', 'Clusters']],
        on='ID',
        how='left'
    )

# ============ BƯỚC 2: CHUẨN BỊ DỮ LIỆU ============
df_model = df_temp_1.copy()

features = [
    'Fre#visit',
    'NeedstateGroup',
    'Companion#group',
    'Daypart',
    'NPS#P3M',
    'Clusters'
]

# Mã hóa biến phân loại
for col in ['NeedstateGroup', 'Companion#group', 'Daypart', 'NPS#P3M']:
    df_model[col] = LabelEncoder().fit_transform(df_model[col].astype(str))

df_model['Fre#visit'] = pd.to_numeric(df_model['Fre#visit'], errors='coerce')
if df_model['Clusters'].dtype == 'O':
    df_model['Clusters'] = LabelEncoder().fit_transform(df_model['Clusters'])

# Xoá dòng thiếu dữ liệu
df_model = df_model.dropna(subset=features + ['churn_flag'])

X = df_model[features]
y = df_model['churn_flag']

# ============ BƯỚC 3: CHIA DỮ LIỆU ============
X_train, X_test, y_train, y_test = train_test_split(
    X, y, stratify=y, test_size=0.2, random_state=42
)

# ============ BƯỚC 4: TRAIN MODELS ============
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

rf = RandomForestClassifier(n_estimators=200, max_depth=10, random_state=42, class_weight='balanced')
lr = LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42)
xgb = XGBClassifier(n_estimators=200, max_depth=5, learning_rate=0.1, use_label_encoder=False, eval_metric='logloss')
lgbm = LGBMClassifier(n_estimators=200, max_depth=5, random_state=42, verbose=-1)

rf.fit(X_train, y_train)
lr.fit(X_train, y_train)
xgb.fit(X_train, y_train)
lgbm.fit(X_train, y_train)

models = {
    'Random Forest': (rf, rf.predict(X_test), rf.predict_proba(X_test)[:, 1]),
    'Logistic Regression': (lr, lr.predict(X_test), lr.predict_proba(X_test)[:, 1]),
    'XGBoost': (xgb, xgb.predict(X_test), xgb.predict_proba(X_test)[:, 1]),
    'LightGBM': (lgbm, lgbm.predict(X_test), lgbm.predict_proba(X_test)[:, 1])
}

# Mapping mô hình sang bảng màu heatmap (cmap)
cmap_map = {
    'Random Forest': 'Greens',
    'Logistic Regression': 'Blues',
    'XGBoost': 'Oranges',
    'LightGBM': 'Purples'
}

roc_results = {}

for name, (model, pred, proba) in models.items():
    print(f"\n--- {name} ---")
    auc = roc_auc_score(y_test, proba)
    roc_results[name] = auc
    print(f"ROC-AUC: {auc:.4f}")
    print("Classification Report:\n", classification_report(y_test, pred))

    # Vẽ confusion matrix với màu riêng
    cm = confusion_matrix(y_test, pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap=cmap_map.get(name, 'Blues'))
    plt.title(f"Confusion Matrix - {name}")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()

# ============ BƯỚC 5: VẼ BIỂU ĐỒ ROC-AUC ============
colors = {
    'Random Forest': 'green',
    'Logistic Regression': 'blue',
    'XGBoost': 'orange',
    'LightGBM': 'purple'
}

plt.figure(figsize=(8, 5))
plt.bar(roc_results.keys(), roc_results.values(), color=[colors[k] for k in roc_results.keys()])
plt.title("ROC-AUC Comparison by Model")
plt.ylabel("ROC-AUC Score")
plt.ylim(0.5, 1.0)
for i, v in enumerate(roc_results.values()):
    plt.text(i, v + 0.01, f"{v:.3f}", ha='center', fontweight='bold')
plt.show()
```

```python
def churn_risk_group(proba):
    if proba >= 0.9:
        return 'Nguy cơ rất cao'
    elif proba >= 0.6:
        return 'Nguy cơ cao'
    elif proba >= 0.3:
        return 'Nguy cơ trung bình'
    else:
        return 'Nguy cơ thấp'

df_model['risk_group'] = df_model['churn_proba'].apply(churn_risk_group)
```
